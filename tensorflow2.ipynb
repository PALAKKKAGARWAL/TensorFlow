#### Changing the datatype of a tensor.

```python
import tensorflow as tf
```

```python
B=tf.constant([1.7,7.4])
B.dtype
```

```python
C=tf.constant([7,10])
C.dtype
```

```python
#### To change from float32 to float16 (reduced precision)
B=tf.cast(B, dtype=tf.float16)
B.dtype
```

```python
D=tf.cast(B,dtype=tf.float16)
D, D.dtype
```

```python
#### Int 32 to float 32
E=tf.cast(C,dtype=tf.float32)
E.dtype
```

#### Aggregating Tensors= condensing then from multiple values down to a smaller amount of values.

```python
## Get the absolute values.
D=tf.constant([-7,-10])
D
tf.abs(D)
```

#### Lets go through the following forms of aggregation:
1) get the min
2) get the max
3) get the mean of a tensor
4) get the sum of a tensor

```python
import numpy as np
```

```python
E=tf.constant(np.random.randint(0,100,size=50))
E
```

```python
## Find the min
tf.reduce_min(E)
```

```python
np.min(E)
```

```python
tf.reduce_max(E)
```

```python
tf.reduce_mean(E)
```

```python
tf.reduce_sum(E)
```
