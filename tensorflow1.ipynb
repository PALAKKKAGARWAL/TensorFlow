### Introduction to Tensors

```python
import tensorflow as tf
print(tf.__version__,)
```

```python
# Creating tensors with tf.constant()
scalar=tf.constant(7)
scalar
```

```python
## Check the number of dimensions of a tensor
scalar.ndim
```

```python
## Create a vector
vector=tf.constant([10,10])
vector
```

```python
vector.ndim
```

```python
## CReate a matrix
matrix=tf.constant([[10,7],[7,10]])
matrix
```

```python
matrix.ndim
```

```python
## Another matrix
another_matrix=tf.constant([[10.,7.],[6.,7.],[8.,9.]],dtype=tf.float16) # specify the data type with dtype parameter.
another_matrix
# using dtype =16 saves space and memory.
```

```python
another_matrix.ndim ## No of elements in shape.
```

```python
## Lets create tensor
tensor=tf.constant([[[1,2,3,],[4,5,5]],[[6,7,8],[11,22,33]],[[12,13,14],[13,23,34]]])
tensor
```

```python
tensor.ndim
```

##### Creating tensors with tf.variable

```python
## Create the same tensor with tf.Variable() as above
changeable_tensor=tf.Variable([10,7])
unchangeable_tensor=tf.constant([10,7])
changeable_tensor,unchangeable_tensor
```

```python
## To change
changeable_tensor[0]=7
```

```python
changeable_tensor[0].assign(7)
changeable_tensor
```

```python
unchangeable_tensor[0].assign(7)
unchangeable_tensor
```

```python
#### Creating Random Tensors.
#### Random tensors are tensors of some abitrary size which contain random numbers.

random_1=tf.random.Generator.from_seed(7) #set seed for reproductivity
random_1=random_1.normal(shape=(3,2))
random_1
```

```python
random_2=tf.random.Generator.from_seed(7)
random_2=random_2.normal(shape=(3,2))
random_2
```

```python
## Are they equal?
random_1,random_2,random_1==random_2
```

```python
## Shufffling the order of elements in a tensor.
## Shuffle the tensor (valuable for when you want to shuffle your data so the inherent order doesn't effect learning.)
not_shuffled=tf.constant([[10,7],
[3,4],[2,5]])
tf.random.shuffle(not_shuffled)
```

```python
tf.random.set_seed(42) ## set the behaviour
tf.random.shuffle(not_shuffled,seed=42)
```
